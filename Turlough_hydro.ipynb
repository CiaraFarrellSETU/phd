```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Prepare dataframe with datetime
df_seasonal = df.copy()
df_seasonal['Date'] = pd.to_datetime(df_seasonal.iloc[:, 0])
df_seasonal['Month'] = df_seasonal['Date'].dt.month

# Map months to seasons
def get_season(month):
    if month in [12, 1, 2]:
        return 'Winter'
    elif month in [3, 4, 5]:
        return 'Spring'
    elif month in [6, 7, 8]:
        return 'Summer'
    else:
        return 'Autumn'

df_seasonal['Season'] = df_seasonal['Month'].apply(get_season)

# Create subplots
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# 1. Monthly Seasonality (Boxplot)
sns.boxplot(
    ax=axes[0],
    x='Month',
    y='042LGHCOY',
    data=df_seasonal,
    hue='Month',
    palette='Blues',
    legend=False
)
axes[0].set_title('Monthly Water Levels: Lough Coy (042LGHCOY)')
axes[0].set_xlabel('Month')
axes[0].set_ylabel('Recorded Value')
axes[0].set_xticks(range(12))
axes[0].set_xticklabels(['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'])
axes[0].grid(True, linestyle='--', alpha=0.5)

# 2. Seasonal Average (Barplot)
season_order = ['Spring', 'Summer', 'Autumn', 'Winter']
sns.barplot(
    ax=axes[1],
    x='Season',
    y='042LGHCOY',
    data=df_seasonal,
    order=season_order,
    hue='Season',
    palette='crest',
    legend=False,
    errorbar=None
)
axes[1].set_title('Average Water Levels by Season: Lough Coy (042LGHCOY)')
axes[1].set_xlabel('Season')
axes[1].set_ylabel('Average Value')
axes[1].grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

### How to Interpret the Correlation Matrix Heatmap

The value in each cell of the matrix is the **Pearson Correlation Coefficient**, which measures the linear relationship between the water levels of two sites over time. The values range from **-1.0 to +1.0**:

* **Near +1.0 (Strong Positive Correlation / Deep Red):** The water levels at both sites rise and fall in near-perfect synchronization. This suggests they share the same hydrology, are close geographically, or belong to the same aquifer/catchment basin.
* **Near 0.0 (No Correlation / Light Orange/White):** There is no clear linear relationship between the water levels of the two sites. They behave independently of each other.
* **Near -1.0 (Strong Negative Correlation / Deep Blue):** One site's water level rises when the other falls. (Note: In this groundwater/surface water dataset, we don't see negative values, meaning no sites behave in opposite directions).

#### Key Findings from your Heatmap:

1. **Extremely High Synchronization ($r \ge 0.95$):**
   * **042LGHCOY** (Lough Coy) and **043CAHGLS** (Caherglassaun) have a correlation of **0.95**.
   * **043CAHGLS** (Caherglassaun) and **058LGEALN** (Lough Gealain) have a correlation of **0.96**.
   * *Interpretation:* These sites are almost certainly connected hydrologically (likely part of the same active turlough or karst conduit system in County Galway) and respond identically to rainfall events.

2. **High General Synchronization ($0.80 \le r < 0.95$):**
   * Most sites share a correlation above **0.70/0.80**.
   * *Interpretation:* This indicates a dominant regional climate signal. Because all these sites are in Ireland, they all experience similar seasonal recharge (heavy winter rain filling the lakes) and summer recession (drying out in summer), driving up their general correlation even if they aren't directly connected underground.

3. **Relatively Lower Correlation ($r < 0.50$):**
   * **074COLGAG** (Colgagh Lough) has a lower correlation with **727CSTLEL** (Castle Lough) (**0.38**) and **051TERMNS** (Termon South) (**0.38**).
   * *Interpretation:* While they still trend in the same broad seasonal direction (no negative correlations), their local hydrological behavior, soil geology, or spring feeds likely differ significantly from each other.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Ensure Date is datetime
df_flash = df.copy()
df_flash['Date'] = pd.to_datetime(df_flash.iloc[:, 0])
df_flash = df_flash.sort_values('Date').reset_index(drop=True)

# Calculate daily differences (rate of change) for the top 15 sites
# We take the absolute value of the day-to-day change
rate_of_change = df_flash[top_15_cols].diff().abs()

# Calculate the average daily absolute change (flashiness index) for each site
flashiness = rate_of_change.mean().sort_values(ascending=False).reset_index()
flashiness.columns = ['Site', 'Mean Absolute Daily Change']

# Plot the Flashiness Index
plt.figure(figsize=(12, 6))
sns.barplot(
    x='Mean Absolute Daily Change',
    y='Site',
    data=flashiness,
    hue='Site',
    palette='flare',
    legend=False
)
plt.title('Hydrological "Flashiness": Mean Absolute Daily Water Level Change', fontsize=14, fontweight='bold')
plt.xlabel('Mean Absolute Daily Change (Meters/Units per Day)')
plt.ylabel('Monitoring Site')
plt.grid(True, linestyle='--', alpha=0.4)
plt.tight_layout()
plt.show()

# Display the sorted dataframe
display(flashiness)
```

```python
import pandas as pd

# Prepare dataframe with Datetime
df_extremes_ind = df.copy()
df_extremes_ind['Date'] = pd.to_datetime(df_extremes_ind.iloc[:, 0])

extreme_records = []

for col in top_15_cols:
    # Drop NaNs to find true valid records
    valid_series = df_extremes_ind[['Date', col]].dropna()
    if not valid_series.empty:
        # Find Max
        max_idx = valid_series[col].idxmax()
        max_row = valid_series.loc[max_idx]

        # Find Min
        min_idx = valid_series[col].idxmin()
        min_row = valid_series.loc[min_idx]

        extreme_records.append({
            'Site': col,
            'Max Date': max_row['Date'].strftime('%Y-%m-%d'),
            'Max Level': round(max_row[col], 3),
            'Min Date': min_row['Date'].strftime('%Y-%m-%d'),
            'Min Level': round(min_row[col], 3)
        })

extremes_df = pd.DataFrame(extreme_records)
display(extremes_df)
```

```python
import pandas as pd

# Prepare data with Datetime and Year
df_yearly_ext = df.copy()
df_yearly_ext['Date'] = pd.to_datetime(df_yearly_ext.iloc[:, 0])
df_yearly_ext['Year'] = df_yearly_ext['Date'].dt.year

yearly_extreme_records = []

# Loop through each site and each unique year in the dataset
for col in top_15_cols:
    for year in sorted(df_yearly_ext['Year'].unique()):
        # Filter data for this specific site and year, dropping NaNs
        site_year_data = df_yearly_ext[df_yearly_ext['Year'] == year][['Date', col]].dropna()

        if not site_year_data.empty:
            # Find Max for this year
            max_idx = site_year_data[col].idxmax()
            max_row = site_year_data.loc[max_idx]

            # Find Min for this year
            min_idx = site_year_data[col].idxmin()
            min_row = site_year_data.loc[min_idx]

            yearly_extreme_records.append({
                'Site': col,
                'Year': year,
                'Max Date': max_row['Date'].strftime('%Y-%m-%d'),
                'Max Level': round(max_row[col], 3),
                'Min Date': min_row['Date'].strftime('%Y-%m-%d'),
                'Min Level': round(min_row[col], 3)
            })

# Convert to DataFrame and display
yearly_extremes_df = pd.DataFrame(yearly_extreme_records)
display(yearly_extremes_df)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Calculate the annual range (fluctuation buffer) for each site per year
yearly_ranges = yearly_extremes_df.copy()
yearly_ranges['Annual Range'] = yearly_ranges['Max Level'] - yearly_ranges['Min Level']

# Pivot the data to create a clean heatmap of fluctuation ranges (Years vs. Sites)
range_pivot = yearly_ranges.pivot(index='Site', columns='Year', values='Annual Range')

# Plot a heatmap of annual fluctuations
plt.figure(figsize=(14, 8))
sns.heatmap(
    range_pivot,
    annot=True,
    fmt=".2f",
    cmap='YlOrRd',
    linewidths=0.5,
    cbar_kws={'label': 'Annual Fluctuation Range (Meters/Units)'}
)
plt.title('Annual Water Level Fluctuation Range (Max - Min) per Site', fontsize=14, fontweight='bold')
plt.xlabel('Year', fontsize=12)
plt.ylabel('Monitoring Site', fontsize=12)
plt.tight_layout()
plt.show()

# Identify the most volatile years on average across all top 15 sites
average_annual_range = yearly_ranges.groupby('Year')['Annual Range'].mean().reset_index()
print("=== AVERAGE HYDROLOGICAL SWING RANGE BY YEAR ===")
print(average_annual_range.to_string(index=False))
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

# 1. Re-calculate Flashiness (mean absolute daily change)
df_flash = df.copy()
df_flash['Date'] = pd.to_datetime(df_flash.iloc[:, 0])
df_flash = df_flash.sort_values('Date').reset_index(drop=True)
flashiness = df_flash[top_15_cols].diff().abs().mean()

# 2. Re-calculate Mean Annual Fluctuation Range
yearly_ranges = yearly_extremes_df.copy()
yearly_ranges['Annual Range'] = yearly_ranges['Max Level'] - yearly_ranges['Min Level']
mean_annual_range = yearly_ranges.groupby('Site')['Annual Range'].mean()

# Combine features into a single DataFrame
features_df = pd.DataFrame({
    'Site': top_15_cols,
    'Flashiness': [flashiness[col] for col in top_15_cols],
    'Mean Annual Range': [mean_annual_range[col] for col in top_15_cols]
})

# 3. Standardize the features for K-Means
scaler = StandardScaler()
scaled_features = scaler.fit_transform(features_df[['Flashiness', 'Mean Annual Range']])

# Run K-Means Clustering
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
features_df['Cluster'] = kmeans.fit_predict(scaled_features)

# Order clusters logically by Mean Annual Range to make classification names consistent
idx_sorted_by_range = features_df.groupby('Cluster')['Mean Annual Range'].mean().sort_values().index
cluster_mapping = {
    idx_sorted_by_range[0]: 'Highly Buffered (Groundwater/Deep Lakes)',
    idx_sorted_by_range[1]: 'Moderately Fluctuating',
    idx_sorted_by_range[2]: 'Highly Dynamic (Karst/Turloughs)'
}
features_df['Category'] = features_df['Cluster'].map(cluster_mapping)

# 4. Visualize the Clusters
plt.figure(figsize=(12, 8))
sns.scatterplot(
    x='Flashiness',
    y='Mean Annual Range',
    hue='Category',
    style='Category',
    data=features_df,
    palette='Set1',
    s=150,
    edgecolor='black'
)

# Annotate the points with Site IDs
for i, row in features_df.iterrows():
    plt.text(
        row['Flashiness'] + 0.002,
        row['Mean Annual Range'] + 0.1,
        row['Site'],
        fontsize=10,
        weight='bold' if row['Category'] == 'Highly Dynamic (Karst/Turloughs)' else 'normal'
    )

plt.title('K-Means Clustering of Monitoring Sites based on Hydrological Behavior', fontsize=14, fontweight='bold')
plt.xlabel('Flashiness (Mean Absolute Daily Delta, meters/day)', fontsize=12)
plt.ylabel('Mean Annual Fluctuation Range (meters)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend(title='Hydrological Category', fontsize=11, title_fontsize=12, loc='upper left')
plt.tight_layout()
plt.show()

# Display the final clustered DataFrame
display(features_df.sort_values(by='Mean Annual Range', ascending=False))
```

A Negative Value (Blue cells): The Base Site (Row) reacts BEFORE the Compared Site (Column). The base site leads, suggesting it is physically upstream or receives water faster.
A Positive Value (Red cells): The Base Site (Row) reacts AFTER the Compared Site (Column). The base site lags, indicating it is downstream or has slower hydrologic transmission.
White/Near Zero (0): The sites are in immediate, near-perfect synchronization with zero delay.
